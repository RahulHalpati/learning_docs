```python
from enum import Enum
app = FastAPI()

class Kind(str, Enum):
    text = "text"
    image = "image"

@app.get("/")
def call(kind:Kind):
    print(Kind)
    return {"kind":Kind}

@app.get("/files/{path:path}")
def finalpath(path: str):
    return {"path":path}

testing = TestClient(app)
testing.get("/").text
testing.get("/files/a/b/c.txt").text
```
